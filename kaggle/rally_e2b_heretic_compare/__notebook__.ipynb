```python
import os, subprocess, sys
from pathlib import Path
REPO_DIR = Path('/kaggle/working/heretic-to-onnx')
sys.path.insert(0, str(REPO_DIR))
from scripts.kaggle_rally_artifacts import ensure_stage_a_merged, find_artifacts
from scripts.kaggle_rally_e2b_two_stage_export import _merge_scaled

artifact_name = os.environ.get('RALLY_TWO_STAGE_ARTIFACT_NAME', 'rally-e2b-two-stage-sft')
direct_id = os.environ.get('RALLY_DIRECT_MODEL_ID', 'p-e-w/gemma-4-E2B-it-heretic-ara')
base_id = os.environ.get('RALLY_BASE_MODEL_ID', 'google/gemma-4-E2B-it')
work_dir = Path('/kaggle/working/rally-e2b-heretic-compare')
scratch = work_dir / 'scratch'
scratch.mkdir(parents=True, exist_ok=True)

model_specs = [f'base:{base_id}', f'heretic:{direct_id}']
try:
    artifacts = find_artifacts(os.environ.get('RALLY_ARTIFACT_DIR', ''), artifact_name)
    stage_a = ensure_stage_a_merged(artifacts, base_model_id=direct_id, scratch_dir=scratch)
    rp_merged = work_dir / 'a100-b75-merged'
    if rp_merged.exists():
        import shutil
        shutil.rmtree(rp_merged, ignore_errors=True)
    _merge_scaled(stage_a, artifacts / 'stage-b-adapter', rp_merged, float(os.environ.get('RALLY_STAGE_B_SCALE', '0.75')))
    model_specs.append(f'rp-a100-b75:{rp_merged}')
    print('rp_artifact_dir=', artifacts)
except Exception as exc:
    print('rp_merge_skipped=', type(exc).__name__, exc)

cmd = [
    sys.executable, str(REPO_DIR / 'scripts/kaggle_rally_model_compare_scorecard.py'),
    '--work-dir', str(work_dir),
    '--report-path', str(work_dir / 'rally-e2b-heretic-compare-report.json'),
    '--models', ','.join(model_specs),
    '--refusal-probe-count', os.environ.get('RALLY_REFUSAL_PROBE_COUNT', '100'),
]
subprocess.check_call(cmd)
```
